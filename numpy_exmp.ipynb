```python
# ================================
# TASK 1
# Import Libraries and Load Dataset
# ================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load Dataset
df = pd.read_csv("gender_submission.csv")

# Display first 5 rows
print("HEAD OF DATASET")
print(df.head())

# Display last 5 rows
print("\nTAIL OF DATASET")
print(df.tail())

# Display shape
print("\nSHAPE OF DATASET")
print(df.shape)

# Display column names
print("\nCOLUMN NAMES")
print(df.columns)


# ================================
# TASK 2
# Check Datatypes, Summary, Missing Values
# ================================

# Datatypes
print("\nDATA TYPES")
print(df.dtypes)

# Summary Statistics
print("\nSUMMARY STATISTICS")
print(df.describe())

# Missing Values
print("\nMISSING VALUES")
print(df.isnull().sum())


# ================================
# TASK 3
# Handle Missing Values, Remove Duplicates,
# Rename Columns
# ================================

# Fill missing numeric values with mean
numeric_cols = df.select_dtypes(include=np.number).columns

for col in numeric_cols:
    df[col] = df[col].fillna(df[col].mean())

# Remove duplicates
df = df.drop_duplicates()

# Rename columns if necessary
df.rename(columns={
    'PassengerId': 'Passenger_ID',
    'Survived': 'Survival_Status'
}, inplace=True)

# Final dataset info
print("\nUPDATED DATASET")
print(df.head())

print("\nUPDATED COLUMN NAMES")
print(df.columns)

print("\nFINAL SHAPE")
# print(df.shape())


# ================================
# TASK 4
# Data Visualization using Matplotlib
# ================================

# -----------------------------
# LINE PLOT
# -----------------------------
plt.figure(figsize=(8,5))

plt.plot(df.index, df['Passenger_ID'])

plt.title("Line Plot of Passenger IDs")
plt.xlabel("Index")
plt.ylabel("Passenger ID")

plt.show()


# -----------------------------
# BAR CHART
# -----------------------------
plt.figure(figsize=(8,5))

survival_count = df['Survival_Status'].value_counts()

plt.bar(
    survival_count.index.astype(str),
    survival_count.values
)

plt.title("Bar Chart of Survival Status")
plt.xlabel("Survival Status")
plt.ylabel("Count")

plt.show()


# -----------------------------
# PIE CHART
# -----------------------------
plt.figure(figsize=(7,7))

plt.pie(
    survival_count.values,
    labels=['Not Survived', 'Survived'],
    autopct='%1.1f%%'
)

plt.title("Pie Chart of Survival Distribution")

plt.show()


# ================================
# UNIVARIATE ANALYSIS
# ================================

# -----------------------------
# HISTOGRAM
# -----------------------------
plt.figure(figsize=(8,5))

plt.hist(df['Passenger_ID'], bins=10)

plt.title("Histogram of Passenger IDs")
plt.xlabel("Passenger ID")
plt.ylabel("Frequency")

plt.show()


# -----------------------------
# BOXPLOT
# -----------------------------
plt.figure(figsize=(6,5))

plt.boxplot(df['Passenger_ID'])

plt.title("Boxplot of Passenger IDs")

plt.show()
```

# An agent / machine is said to learn from experience with respect to some class of tasks (t) and performance measure (P), if its performance at tasks in T, as measured by P, improves with experience (E).

#                                                                                             :- Tom M. Mitchell(1997)



Understanding T,P and E in machine learning

1. Clarity of learning objective

    Defining task (T) helps specify what the sysyem is supported to learn or do
    Example: Classify emails as spams or not spams

2. Measuring improvements

    Defining performance(P) gives a quantitative way to measure success or programs
    ex: Accuracy , error rate, F1 score etc...

3. Understanding the learning source

    Defining experience(E) identifies how the system gains knowledge
    ex: past data, feedback , user interactions

4. Ensure Scientific Evaluation

    Makes it possible to compare diff. learning system objectively
